```python
import numpy as np
```

```python
print(np.array([1,2]) + np.array([3,4]))
print(np.array([1,2]) * 3)
```

```python
class Vector:
    def __init__(self, *components):
        self.v = np.array(components)

    def __add__(self, other):
        return Vector(*np.add(self.v, other.v))

    def __eq__(self, other):
        return np.array_equal(self.v, other.v)

    def __repr__(self):
        return f"Vector{self.v}"

    def dot(self, other):
        return self.v.dot(other.v)

    def norm(self):
        return np.linalg.norm(self.v)

    def __len__(self):
        return len(self.v)

    def __mul__(self, other):
        return Vector(*self.v * other)

print(Vector(1, 2))
assert Vector(1, 2) + Vector(3,4) == Vector(4, 6)
assert Vector(1, 2).dot(Vector(3,4)) == 11
assert np.isclose(Vector(0.1, 0.2).norm(), 0.05 ** 0.5)
assert Vector(3,4).norm() == 5
assert len(Vector(1,2)) == 2
assert Vector(1, 2) * 3  == Vector(3, 6)
```

```python
print(0.1 + 0.2)                        # 0.30000000000000004
assert 0.1 + 0.2 != 0.3                 # plain == fails on floats
assert np.isclose(0.1 + 0.2, 0.3)       # isclose allows for tiny rounding errors
```
